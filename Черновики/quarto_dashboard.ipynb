## Строка 2 {height=50%}

```python
#| include: False

import plotly.graph_objects as go
import pandas as pd
import numpy as np

import plotly.io as pio
pio.renderers.default = 'notebook_connected'

# Генерация данных
np.random.seed(42)
categories = ['Продукты', 'Транспорт', 'Жилье', 'Развлечения', 'Здоровье', 'Образование']
values = np.random.randint(100, 500, size=len(categories))

# Создание DataFrame
df = pd.DataFrame({
    'Категория': categories,
    'Сумма': values
})

# Расчет процентов
total = df['Сумма'].sum()
df['Процент'] = (df['Сумма'] / total * 100).round(1)
df['Метка'] = df['Категория'] + '<br>' + df['Сумма'].astype(str) + ' руб.<br>' + df['Процент'].astype(str) + '%'
df
```

```python
# Построение круговой диаграммы с выносками
fig = go.Figure(data=[go.Pie(
    labels=df['Категория'],
    values=df['Сумма'],
    text=df['Метка'],  # Текст с выносками
    textposition='outside',  # Выноски снаружи
    textinfo='text',  # Показываем кастомный текст
    hoverinfo='label+value+percent',
    marker=dict(
        colors=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7', '#DDA0DD'],
        line=dict(color='white', width=0)
    ),
    hole=0.5,
    showlegend=True,
    #pull=[0, 0.08, 0, 0.08, 0],  # Фиксированное выделение секторов
)])

fig.update_layout(
    title=dict(
        text='Распределение расходов по категориям',
        font=dict(size=20, family='Arial', color='#2c3e50'),
        x=0.5
    ),
    width=800,
    height=600,
    font=dict(size=14, family='Arial'),
    margin=dict(l=50, r=50, t=80, b=50)
)

fig.show()
```
