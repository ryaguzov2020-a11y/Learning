```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px

# Example dataset with a categorical column
df = pd.DataFrame({
    "category": ["A", "A", "B", "B", "C", "C"],
    "x": [1, 2, 1, 2, 1, 2],
    "y": [10, 15, 12, 18, 9, 11]
})

# Unique category values
cats = df["category"].unique()

# Create one trace per category
fig = go.Figure()

for cat in cats:
    sub = df[df["category"] == cat]
    fig.add_trace(
        go.Scatter(
            x=sub["x"],
            y=sub["y"],
            mode="lines+markers",
            name=str(cat),
            visible=True if cat == cats[0] else False  # initially show first category
        )
    )

# Build dropdown buttons
buttons = []

for i, cat in enumerate(cats):
    # visible mask: only trace i visible
    visible_mask = [False] * len(cats)
    visible_mask[i] = True

    buttons.append(
        dict(
            label=str(cat),
            method="update",
            args=[
                {"visible": visible_mask},
                {"title": f"Category: {cat}"}
            ]
        )
    )

# Add "All" button
buttons.insert(
    0,
    dict(
        label="All",
        method="update",
        args=[
            {"visible": [True] * len(cats)},
            {"title": "All Categories"}
        ]
    )
)

# Add dropdown to layout
fig.update_layout(
    title="Category Filter Example",
    updatemenus=[
        dict(
            buttons=buttons,
            showactive=True,
            direction="down",
            x=0.0,
            y=1.15
        )
    ]
)

fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go

df = pd.DataFrame({
    'Склад': ['Склад А', 'Склад А', 'Склад Б', 'Склад Б', 'Склад В', 'Склад В'],
    'ГодМесяц': ['2024-01', '2024-02', '2024-01', '2024-02', '2024-01', '2024-02'],
    'Сумма_Кт': [1000, 1200, 800, 900, 1500, 1400],
    'кластер': ['Север', 'Север', 'Север', 'Север', 'Юг', 'Юг']
})

# Сводная таблица
pivot = df.pivot_table(index='ГодМесяц', columns='Склад', values='Сумма_Кт', aggfunc='sum').fillna(0)

# Словарь: какие склады в каком кластере
cluster_warehouses = {
    'Север': ['Склад А', 'Склад Б'],
    'Юг': ['Склад В']
}

fig = go.Figure()

# Добавляем линии для всех складов сразу
for warehouse in pivot.columns:
    fig.add_trace(go.Scatter(
        x=pivot.index,
        y=pivot[warehouse],
        mode='lines+markers',
        name=warehouse
    ))

# Функция видимости для кластера
def get_visibility(cluster):
    wh_list = cluster_warehouses[cluster]
    return [wh in wh_list for wh in pivot.columns]

# Создаем кнопки
buttons = [
    dict(
        label='Все',
        method='update',
        args=[{'visible': [True]*len(pivot.columns)}]
    ),
    dict(
        label='Север',
        method='update',
        args=[{'visible': get_visibility('Север')}]
    ),
    dict(
        label='Юг',
        method='update',
        args=[{'visible': get_visibility('Юг')}]
    )
]

fig.update_layout(
    updatemenus=[dict(buttons=buttons, direction='down', x=1.1, y=1.15)],
    title='Сумма_Кт по складам'
)

fig.show()
```
